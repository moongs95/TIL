# 자료구조와 라이브러리

- 알고리즘이 요리의 과정이라면? 자료구조는 조리 도구!
- Python은 다른 언어 대비 기본 자료형으로 제공되는 자료구조가 많음
- 패키지 import가 필요한 자료구조만을 다룸

## Queue
- 선입 선출 원칙을 따르는 자료구조
- queue.Queue를 이용해 사용 가능

## Priority Queue
- Comparator에 따라 우선순위가 결정되는 자료구조
- Heap 자료 구조를 기반으로 구현 됨
- queue.PriorityQueue()를 이용해 사용 가능

- put(item): 큐에 요소 추가
- get(): 큐에서 요소 제거 및 반환
- qsize(): 큐의 요소 개수 반환
- empty(): 큐가 비어 있는지 확인
- full(): 큐가 꽉 찼는지 확인(maxsize를 지정한 경우)

### 실습

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __lt__(self, other):
        return self.age < other.age
```

```python
import queue

q = queue.Queue()
pq = queue.PriorityQueue()

# q.put
q.put(1)
q.put(3)
q.put(5)

# pq.put
pq.put(User("a", 20))
pq.put(User("b", 40))
pq.put(User("c", 30))
```

```python
# q.get
for i in range(q.qsize()):
    print(q.get())
```

```python
# pq.get
for i in range(pq.qsize()):
    u = pq.get()
    print(f"{u.name}:{u.age}")
```
